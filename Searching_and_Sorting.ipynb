#LINEAR SEARCH

###enumerate() is a build-in python function that lets you loop over a list , while also keep track of index of element.

### Basic linear search using loop

```python
def linear_search(arr,target):
  for i in range(len(arr)):
    if arr[i] == target:
      return i
  return -1

arr = [2,5,3,6,71,9,1,2]
print(linear_search(arr,2))
```

### Using enumerate() --> Best way

```python
def lnr_search(arr, target):
  for index,value in enumerate(arr):
    if value == target:
      return index
  return -1

arr = [2,4,5,1,7,8,6]
print(lnr_search(arr,8))
```

### Linear Search using recursion

```python
def linear_srch(arr,target,index = 0):
  if index >= len(arr):
    return -1
  if arr[index] == target:
    return index
  return linear_srch(arr,target, index+1)

arr = [3,2,1,5,7,8,9]
print(linear_srch(arr,1))
```

## Find all occurances of the element

```python
# In this code index of all dublicate value are shown
#Instead of stopping early, we check the entire array.

def all_occurances(arr,target):
  result = [ ]
  for i,val in enumerate(arr):
    if val == target:
      result.append(i)
  return result

arr = [4,3,2,5,67,2,1]
print(all_occurances(arr,5))
```

## Find first and last occurance
### First occurance --> store once
### Second occurance --> Keep updating

```python
def first_last(arr,target):
  first = last = -1
  for i, val in enumerate(arr):
    if val == target:
      if first == -1:
        first = i
      last = i
  return first, last

arr = [2,6,2,9,5,3,9,4]
print(first_last(arr,9))
```

#Count Occurance

```python
def count_occurances(arr,target):
  return sum(1 for value in arr if value == target)

arr = [2,6,2,9,5,3,9,4]
print(count_occurances(arr,3))
```

#Find Minimum value

```python
def min_value(arr):
  min_value = arr[0]
  for x in arr:
    if x < min_value:
      min_value = x
  return min_value

arr = [2,6,2,9,5,3,9,4]
print(min_value(arr))
```

# Return first and last position(Leetcode-34)

## Approch


*   First Occurance - go left
*   Second Occurance - go right



```python
def searchrange(nums, target):

  def find_first_occurrence():
    low = 0
    high = len(nums) - 1
    ans = -1

    while low <= high:
      mid = (low + high) // 2

      if nums[mid] == target:
        ans = mid # Potential first occurrence, try to find earlier
        high = mid - 1
      elif nums[mid] < target:
        low = mid + 1
      else: # nums[mid] > target
        high = mid - 1
    return ans

  def find_last_occurrence():
    low = 0
    high = len(nums) - 1
    ans = -1

    while low <= high:
      mid = (low + high) //2

      if nums[mid] == target:
        ans = mid # Potential last occurrence, try to find later
        low = mid + 1
      elif nums[mid] < target:
        low = mid + 1
      else: # nums[mid] > target
        high = mid - 1
    return ans

  first_pos = find_first_occurrence()
  last_pos = find_last_occurrence()

  return [first_pos, last_pos]

# Example usage with a sorted array, as binary search requires it.
# Original unsorted: [7,3,4,1,9,4,3,0]
nums = [3, 4, 4, 7, 7, 7, 9, 9]
print(searchrange(nums, 7))  # Output: [3, 5}
```

#Search in rotated sorted array
##Approach


*   check which half is sorted
*   decide which target lies



```python
def rotated(nums,target):
  left = 0
  right = len(nums) - 1

  while left <= right:
    mid = left + (right - left) // 2

    if nums[mid] == target:
      return mid

    if nums[left] <= nums[mid]:
      if nums[left] <=  target < nums[mid]:
        right = mid - 1
      else:
        left = mid + 1

    else:
      if nums[mid] < target <= nums[right]:
        left = mid + 1
      else:
        right = mid - 1

  return -1


nums = [7,2,6,5,0,1,2,3,4]
print(rotated(nums,2))
```

#Square root - (leetcode-69)
##Thinking

*  Anser lies between 1 and x.
*  Binary search in value.

##Approch


*  check mid * mid <= x
  






```python
def sqrt(x):
  low = 1
  high = x
  ans = 0

  while low <= high:
    mid = low + (high - low) // 2

    if mid * mid <= x:
      ans = mid
      low = mid + 1
    else:
      high = mid - 1
  return ans

x = 8
print(sqrt(x))
```
